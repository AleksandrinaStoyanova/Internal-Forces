# Internal Forces

#### Abstract

In the following article we consider the determination of the Internal forces of a simple beam, loaded with distributed and poin loads.

First we use the cut section method  - we cut a small section of the beam, apply all the adjacent external loads on that section and we determine the internal forces by using the equilibrium equations.
In the shown example, the internal forces are determined for a plane loaded beam. 
In this case first we calculate the support reactions and then - the internal forces. Also we calculate the extremum values of the internal forces and visualize their distribution along the beam. 
It's important to mention that for the bending moment diagram, the positive value of the internal forces is on the tense side of the beam (for a simple beam that is on the negative side of the axis of the beam).

Then we consider the differential dependencies between the  internal forces and the external loads, acting on the beam. We achieve this by moving the two sections towards one another.

Finally we examine the determination of the internal forces in case of a complex distribited load. In that case we don't need to determine the support reactions in advance - we get them by integrating the differential dependencies between the internal forces and the external loads. 

### Introduction

The determination of the $\mathbf{Internal\ Forces}$ is one of the most important tasks in structural section sizing (IF's extremum values).
The folowing conclusions and calculations are made presumpting the material is homogeneous, meaning it has identical properties at every point.
This is a major presumpion in the theory of Strength of Materials.

The considered calculations are made for Statically determinated system (the number of unknown support reactions is equal to the number of independent equilibrium equations) and static external forces (constant loads, that do not change in magnitude, direction or position over time).

Let's consider a beam, loaded with external forces $F_i$ that is balanced (fig. 1):





```python
Image(filename="Fig_1.jpg", width=300, height=200)
```

We determine the support reactions of the beam using static equilibrium 
(the sum of external forces $\sum{F_x} = 0,\  \sum{F_y} = 0$ and moments $\sum{M} = 0$ equals zero, holding the structure stationary).
Let's cut the beam at some place, perpendicular to the axis of the beam with a plane $\mathbf{\alpha}$. We 'll call the two parts $\mathbf{left}$ and $\mathbf{right}$.
The plane $\mathbf\alpha$ divides the forces into two components: $\vec{F_L}$ and $\vec{F_R}.\ \vec{F_L}$ includes the forces, acting on the left part of the beam and $\vec{F_R}$ - the forces, acting on the right part. At the point where we have cut the beam, there are an infinite number of internal forces, randomly orientated - $S_i$, with witch the left and the right part of the beam react. $S_L^{R}$ are the internal forces from the right part of the beam, acting on the left part, $S_R^{L}$ are the internal forces from the left part of the beam, acting on the right part.
Acording to Newton's Third Law ("For every action there is an equal and opposite reaction") - $S_L^{R},\ S_R^{L}\sim 0$.
Let's concider the left part of the beam. As we assumed that the beam is in balance, then every part of it will also be balanced. So we can write that $F_L,\ S_L^{R}\sim 0$.
The unknowns from this system are the internal forces $S_L^{R}$. They are an important characteristic for the state of the beam with the given external forces. As the internal forses $S_L^{R}$ are an infinite, randomly orientaded number, we can't calculate them from the equation $F_L,\ S_L^R\sim 0$, but we can calculate their resultant vector $\vec{R}$ and resultant moment $\vec{M}$ for the center of the masses of the cross section.
Then we can write $\vec{R} = - \vec{R_L}\\,\  \vec{M} = - \vec{M_L}\ $ for the left part of the beam and $\vec{R} = \vec{R_R},\ \vec{M} = \vec{M_R}$.
Combining the results from the two parts of the beam we come to:

$$(1)\          \left\{ \begin{array}{cc}\vec{R} = - \vec{R_L} = \vec{R_R} \\ 
\ \vec{M} = - \vec{M_L} = \vec{M_R}\end{array}\right.$$







```python
Image(filename="Fig_2.jpg", width=400, height=200)
```

If we project the resultant vector $\vec{R}$ and the resultant moment $\vec{M}$ on the normal axis of the beam and on the plane $\alpha$, which is perpendicular to the normal axis, we get the following forces and moments:\
$\vec{R} = \vec{N} + \vec{Q} = \vec{N} + \vec{Q_y} + \vec{Q_z} = N\vec{i} + Q_y\vec{j} + Q_z\vec{k}$\
$\vec{M} = \vec{M_t} + \vec{M_f} = \vec{M_t} + \vec{M_{f,y}} + \vec{M_{f,z}} = M_t\vec{i} + M_y\vec{j} + M_z\vec{k}$\
where $\vec{i},\ \vec{j},\ \vec{k}$ are the unit vectors along the axes $ x,\ y,\ z$\
$\vec{N}$ - normal force\
$\vec{Q}$ - tangential (shear) force\
$\vec{M_t}$ - torsional moment\
$\vec{M_f}$ - bending moment

In summary $\vec{N},\ \vec{Q},\ \vec{M_t},\ \vec{M_f}$ are called $\mathbf{Internal\ Forces}$ in the plane $\alpha$ of the beam.


For practical purposes, it's easier to work with the scalar components of $\vec{R}$ and $\vec{M}$ than with the vectors themselfs.\
For this purpose we choose a special coordinate system, where the axis $\mathbf{x}$ matches with the external normal of the cross section\
and the axes $\mathbf{y}$ and $\mathbf{z}$ form a right - handed set of vectors.

$$(2)\begin{equation}
\left\{ 
  \begin{aligned}
    & N = R_x = - R_{L,x} = R_{R_x} - normal\ force \\
    & Q_y = R_y = - R_{L,y} = R_{R,y} - shear\ force\ over\ axis\ Cy \\
    & \ Q_z = R_z = - R_{L,z} = R_{R,z} - shear\ force\ over\ axis\ Cz \\
    & \ M_t = M_x = - M_{L,x} = M_{R,x} - torsional\ moment \\
    & \ M_{f,y} = M_y = - M_{L,y} = M_{R,y} - bending\ moment\ on\ Cy\ axis \\
    & \ M_{f,z} = M_z = - M_{L,z} = M_{R,z} - bending\ moment\ on\ Cz\ axis
  \end{aligned} 
\right.
\end{equation}$$

In reality it's quite common that the axis of the beam lies in the same plane as the external forces.

If we take $\mathbf{xz}$ to be that plane, than $Q_y = 0,\ M_t = 0,\ M_z = 0$.

That kind of load is called $\mathbf{plane\ load}$.


### Calculating the Internal forces in a plane loaded beams

So if we want to determine the Internal forces in random cross section of a beam we follow this steps:
1. Calculate all reaction forces and moments at the supports using static equilibrium equations for the entire structure.
2. We make a cut at the point of interest and consider the left or the right part (the part with the fewer loads for easier calculation).
3. On the chosen part of the beam, we put all external forces and the internal forces, considered with positive directions.
4. We solve the three equlibrium equations to calculate the internal forces:
   
   $\sum{H} = 0\ -\ horizontal\ forces$
   
   
   $\sum{V} = 0\ -\ vertical\ forces$
   
   
   $\sum{M} = 0\ -\ moments\ about\ the\ cut$
   



For the design of a structure it's crutial to find the cross section where the internal forces are with their extremum size.
In practice it's best to display the distribution of the internal forces using diagrams.

1. Defining the beam (it's geometry and loads):

```python
l = 4      # beam length
f = 40     # point load
a = 1      # distance from start to point load (must be between 0 and l)
q = 10     # distributed load, we represent it with it's resultant force, acting in the center of the distributed load
```

2. Drawing a simple beam with loads:

```python
def draw_simple_beam(l, f, a, q):
    plt.figure(figsize=(12,2))
    plt.plot([0, l], [0, 0], color="black", linewidth=5)
    plt.annotate('', xy=(a, 0), xytext=(a, 10),
                arrowprops=dict(facecolor='red', shrink=0.05, width=2))
    plt.text(a, 5, f'{f} kN', ha='right')
    plt.xlim(-1, l + 2)
    plt.ylim(-1, f)
    plt.axis('off')
    x_load = np.linspace(0, l, 100)
    plt.fill_between(x_load, 0, q, color='blue', alpha=0.3, label='UDL')
    plt.plot([-1, l], [q, q], color='blue', linewidth=2)
    arrow_x = np.linspace(0, l, 10)
    for x in arrow_x:
        plt.annotate('', xy=(x, 0), xytext=(x, q), 
                     arrowprops=dict(arrowstyle='->', color='blue', lw=1.5))
    plt.text(l/2, 5, f'{q} kN/m', ha='center')

    plt.ylim(-1, q )
    plt.xlim(0, l + 1)
    plt.title("Loads")
    plt.show()
```

```python
draw_simple_beam(l, f, a, q)
```

3. Calculating the reaction forces:

```python
def calculate_spring_reactions(l, f, a, q):
    
    """
    We use that the sum of the moments for each support is 0.
    """
    reaction_1 = q * l / 2 + f * (l - a) / l
    reaction_2 = q * l / 2 + f * a / l
    return [reaction_1, reaction_2]
```

```python
calculate_spring_reactions(l, f, a, q)
```

4. Calculate the internal forces in different cross sections of the beam:


```python
def calculate_bending_moments(x, l, f, a, q):

    """
    For each section along the beam with length x we calculate sum of moments (each force multiplied with
    it's distance to that section).The uniformly distributed load is represented by it's resultant force, 
    applied in the center of the section x.
    """
    
    reaction_1, reaction_2 = calculate_spring_reactions(l, f, a, q)
    bending_moment = reaction_1 * x - (q * x * x) / 2 if x <= a else reaction_1 * x - (q * x * x) / 2 - f * (x - a)
    return bending_moment
```

```python
calculate_bending_moments(1, l, f, a, q)
```

```python
def calculate_shear_forces(x, l, f, a, q):
    
    """
    For each section along the beam with length x we calculate sum of vertical forces.
    The uniformly distributed load is represented by it's resultant force, applied in the center of the section x.
    """
    reaction_1, reaction_2 = calculate_spring_reactions(l, f, a, q)
    shear_force = reaction_1 - q * x  if x <= a else reaction_1 - q * x - f 
    return shear_force
```

```python
calculate_shear_forces(1, l, f, a, q)
```

```python
def vectorize_internal_forces(function):
    v_internal_forces = np.vectorize(function)
    return v_internal_forces
```

5. Find the extremum values of the Internal forces:

```python
x = np.linspace(0, l, 1000)
```

```python
def calculate_extremum_internal_force(function, x, l, f, a, q):
    
    internal_forces = vectorize_internal_forces(function)

    max_internal_force = np.max(internal_forces(x, l, f, a, q))
    min_internal_force = np.min(internal_forces(x, l, f, a, q))
   
    return [max_internal_force, min_internal_force]
```

```python
calculate_extremum_internal_force(calculate_bending_moments, x, l, f, a, q)
```

```python
calculate_extremum_internal_force(calculate_shear_forces, x, l, f, a, q)
```

... and their location:

```python
def calculate_extremum_internal_force_location(function, x, l, f, a, q):

    internal_forces = vectorize_internal_forces(function)

    max_idx_internal_force = np.argmax(internal_forces(x, l, f, a, q))
    max_internal_force_location = x[max_idx_internal_force]

    min_idx_internal_force = np.argmin(internal_forces(x, l, f, a, q))
    min_internal_force_location = x[min_idx_internal_force]

    return [max_internal_force_location, min_internal_force_location]
    
```

```python
calculate_extremum_internal_force_location(calculate_bending_moments, x, l, f, a, q)
```

```python
calculate_extremum_internal_force_location(calculate_shear_forces, x, l, f, a, q)
```

Finding the extremum values of the Internal forces is very important, as based on the result, 
the design of the section is made, and for a reinforced concrete section - the reinforcement is calculated.
The best way to visualize the Internal forces is by drawing their diagrams.
Important thing is that the bending moment diagram is $\textbf{reversed}$ - the positive moments are plotted below the axis of the beam (on the tension side of the beam).

6. Draw moment diagram and calculate bending moment extremum:
   

```python
def draw_internal_forces_diagram(function, l, y_axis_name, force_label, invert_y=True):

    internal_forces = vectorize_internal_forces(function)
    internal_force_values = internal_forces(x, l, f, a, q)
    extremum_internal_force = calculate_extremum_internal_force(function, x, l, f, a, q)
    extremum_internal_force_location = calculate_extremum_internal_force_location(function, x, l, f, a, q)
    
    plt.plot(x, internal_force_values, label=f"{function.__name__}")
    plt.axhline(0, color='black', lw=1)
    plt.fill_between(x, internal_force_values, alpha=0.3)

    plt.scatter(extremum_internal_force_location[0], extremum_internal_force[0], color='red', label=f'{force_label}max={extremum_internal_force[0]:.2f}')
    plt.annotate(f"{force_label}max = {extremum_internal_force[0]:.2f} kNm", 
             xy=(extremum_internal_force_location[0], extremum_internal_force[0]), 
             xytext=(extremum_internal_force_location[0] + 1, extremum_internal_force[0] + 1))
    plt.scatter(extremum_internal_force_location[1], extremum_internal_force[1], color='red', label=f'{force_label}min={extremum_internal_force[1]:.2f}')
    plt.annotate(f"{force_label}min = {extremum_internal_force[1]:.2f} kNm", 
             xy=(extremum_internal_force_location[1], extremum_internal_force[1]), 
             xytext=(extremum_internal_force_location[1] + 0.2, extremum_internal_force[1] + 2))

    if invert_y:
        plt.gca().invert_yaxis()

    diagram_title = function.__name__[10:].replace("_", " ").upper()


    plt.title(diagram_title)
    plt.xlabel("L, m")
    plt.ylabel(y_axis_name)
    plt.grid(True)
    plt.show()
```

```python
draw_internal_forces_diagram(calculate_bending_moments, l, "M, kN.m","M",  True)
```

7. Draw shear diagram and calculate shear force extremum:

```python
draw_internal_forces_diagram(calculate_shear_forces, l, "Q, kN", "Q", False)
```

### Differential dependencies for the Internal forces

Let's concider a straight beam with random loads. We want to find the dependencies between  the Internal forces and the external loads.
To achieve this let's cut a tiny slice of the beam using two sections at $x$ and $x + \Delta{x}$.
We'll concider that the external forces consists of:
 - transverse distributed load with dencity $q(x)$ with two components - $q_y(x)$ and $q_z(x)$
 - axial distributed load with dencity $t(x)$
 - distributed moment with dencity $m(x)$ with it's three components $m_x(x)$, $m_y(x)$ and $m_z(x)$.

```python
Image(filename="Fig_4.jpg", width=400, height=200)
```

Since the two slices are very, very close, the dencity of the loads will not change significantly, so we can concider them as constants.\
If we substitude the distributed loads with their equivalent we'll get:


$$(3)\hspace{60pt} F = q(\xi_1)\Delta{x},\hspace{10pt} T = t(\xi_2)\Delta{x} $$

where $\xi_1$ and $\xi_2$ are some abscissas between $x$ and $x + \Delta{x}$:


$$x < \xi_1 < x + \Delta{x},\hspace{20pt} x < \xi_2 < x + \Delta{x}$$

The force $F$ is applied somewhere between $C$ and $C'$ (fig. 5,  $0 < \lambda < 1$)

```python
Image(filename="Fig_5.jpg", width=400, height=200)
```

The three equations of balance for the tiny cut element are:

$$(4)\hspace{20pt}\begin{align}&\sum{H} = 0\hspace{20pt} N(x + \Delta{x}) - N(x) + T = 0\\
&\sum{V} = 0\hspace{20pt} Q(x + \Delta{x}) - Q(x) + F = 0\\
&\sum{M} = 0\hspace{15pt} M(x + \Delta{x}) - M(x) - Q(x)\Delta{x} + F\lambda\Delta{x} = 0
\end{align}$$


If we substitude the equivalen loads from $(3)$ in the three equations of balance in $(4)$ we get:

$$(5)\hspace{50pt}\begin{align}&\frac{N(x + \Delta{x}) - N(x)}{\Delta{x}} = - t(\xi_2)\\\\
&\frac{Q(x + \Delta{x}) - Q(x)}{\Delta{x}} = - q(\xi_1)\\\\
&\frac{M(x + \Delta{x}) - M(x)}{\Delta{x}} = Q(x) + \lambda{q(\xi_1)}\Delta{x}
\end{align}$$

Let's assume that $\Delta{x}$ is tending to zero, in other words we move section $C'$ towards section $C$.\
The left parts of the equations $(5)$ will tend to the derivatives of the functions $N(x)$, $Q(x)$ and $M(x)$\
and the abscissas $\xi_1$ and $\xi_2$ will tend to $x$. Also in the third equation in $(5)$ as $\Delta{x}\rightarrow 0$, then $\lambda{q(\xi_1)}\Delta{x}\rightarrow 0$.\
Finally for the dependency ot the Internal forces and the external loads in a cross - section we get:

$$(6)\hspace{100pt}\begin{align}&\frac{dN}{dx} = -t(x)\\\\
&\frac{dQ}{dx} = - q(x)\\\\
&\frac{dM}{dx} = Q(x)
\end{align}$$

For randomly situated loads towards the axis of the beam, we have six equations of balance:

```python
Image(filename="Fig_6.jpg", width=500, height=250)
```

$$\begin{align}&\sum{X_i} = 0:\hspace{14pt}N(x + \Delta{x}) - N(x) + t(x).\Delta{x} = 0\\\\
&\sum{Y_i} = 0:\hspace{14pt}Q_y(x + \Delta{x}) - Q_y + q_y(x).\Delta{x} = 0\\\\
&\sum{Z_i} = 0:\hspace{14pt}Q_z(x + \Delta{x}) - Q_z + q_z(x).\Delta{x} = 0\\\\
&\sum{M_{Cx}} = 0:\hspace{5pt}M_t(x + \Delta{x}) - M_t(x) + m_x(x).\Delta{x} = 0\\\\
&\sum{M_{Cy}} = 0:\hspace{5pt}M_y(x + \Delta{x}) - M_y(x) - Q_z(x).\Delta{x} + m_y(x).\Delta{x} + q_z(x).\Delta{x}.\frac{\Delta{x}}{2} = 0\\\\
&\sum{M_{Cz}} = 0:\hspace{5pt}M_z(x + \Delta{x}) - M_z(x) + Q_y(x).\Delta{x} + m_z(x).\Delta{x} - q_y(x).\Delta{x}.\frac{\Delta{x}}{2} = 0\\\\
\end{align}$$

If we divide all six equations by $\Delta{x}$ and since $\Delta{x}\ \rightarrow\ 0$ we can ignore the last terms in the equations for $\sum{M_{Cy}} = 0$ and $\sum{M_{Cz}} = 0$, we come to:


$$\begin{align}
&\frac{N(x + \Delta{x}) - N(x)}{\Delta{x}} = -t(x)\\\\
&\frac{Q_y(x + \Delta{x}) - Q_y(x)}{\Delta{x}} = -q_y(x)\\\\
&\frac{Q_z(x + \Delta{x}) - Q_z(x)}{\Delta{x}} = -q_z(x)\\\\
&\frac{M_t(x + \Delta{x}) - M_t(x)}{\Delta{x}} = -m_x(x)\\\\
&\frac{M_y(x + \Delta{x}) - M_y(x)}{\Delta{x}} = Q_z(x) - m_y(x)\\\\
&\frac{M_z(x + \Delta{x}) - M_z(x)}{\Delta{x}} = - Q_y(x) - m_z(x)\\\\
\end{align}$$

But the left part of the above equations is the definition for a derivative, so we finally get these equations that represent the dependancy between the Internal forces and the external loads:

$$\begin{align}&\frac{dN}{dx} = - t(x)\\\\
&\frac{dQ_y}{dx} = - q_y(x)\\\\
&\frac{dQ_z}{dx} = - q_z(x)\\\\
&\frac{dM_t}{dx} = - m_x(x)\\\\
&\frac{dM_y}{dx} = Q_z(x) - m_y(x)\\\\
&\frac{dM_z}{dx} = - Q_y(x) - m_z(x)\\\\
\end{align}$$

#### Calculating the Spring Reactions by integrating the differential dependancies for the Internal forces

In cases when the external load distribution follows a complex pattern, applying the cross section method is very difficult. In that case it's more convinient to integrate seperatly each one of the equations for the differential dependancies of the Internal forses $equations\ (6)$:

$$(7)\hspace{30pt}\begin{align}&N(x) = - \int t(x) \,dx + C_1\\\\
&Q(x) = - \int q(x) \,dx + C_2\\\\
&M(x) = - \int dx\int q(x) \,dx + C_2 \, x + C_3
\end{align}$$

where $C_1, C_2$ and $C_3$ are the constants of integration, which can be calculated from the conditions for equilibrium.

Let's consider the following example:


We have a simple beam, divided into two sections, the firtst section is loaded with a parabolic distributed load, the second section - with a linear distributed load and there is also a forse F, which acts in the point $C$, where the two distributed loads change:

We perform the integration in $(7)$ for every section.


The load functions are:

$q_1(x) = a_1 \cdot x^2 + b_1\cdot x + c_1\hspace{20pt}$ for the first section
and

$q_2(x) = a_2\cdot x + b_2\hspace{58pt}$ for the second section

If we substitude $\mathbf{x}$ with $\mathbf{0}$ and $\mathbf{l_1 = 2}$ for the start and end of the first section and with $\mathbf{0}$ and $\mathbf{l_2 = 3}$ for the start and end of the second section, we can write the following equations:

$$\begin{align}&q_1(0) = 0\Rightarrow\ c_1 = 0\\\\
&q_1'(0) = 0\Rightarrow\ 2\cdot a_1\cdot x + b_1 = 0\Rightarrow\ b_1 = 0\\\\
&q_1(2) = 24\Rightarrow\ a_1\cdot 2^2 = 24\Rightarrow\ a_1 = 6\Rightarrow\ q_1(x) = 6 \,x^2\\\\
&q_2(0) = 24\Rightarrow\ b_2 = 24\\\\
&q_2(3) = 0\Rightarrow\ a_2 = -8\Rightarrow\ q_2(x) = -8 \,x + 24
\end{align}$$

 - Calculate linear distributed load coefficents:

```python
def calculate_linear_eq_coefficients(start_x, start_y, end_x, end_y):
    
    """
    If we know 2 points of the linear function a * x + b (in our case - the star and the end of the load), 
    we can calculate the coefficients as follows: a = (y_2 - y_1) / (x_2 - x1), b = y_1 - a * x_1
    """
    
    a = (end_y - start_y) / (end_x - start_x)
    b = start_y - a * start_x

    return (a, b)
```

 - Calculate quadratic distributed load coefficients:

```python
def calculate_quadratic_eq_coefficients(f_0, f_end, f_prime_0, end_value):
    c = f_0
    b = f_prime_0
    a = (f_end - b * end_value - c) / (end_value**2)
    return (a, b, c)
```

From the definition of integral we know that $$\int a \,x^{n} \,dx = a\frac{x^{n+1}}{n+1} + C$$

```python
def integrate_function(formula_str, c1=0):

    x = sp.Symbol('x')
    f = sp.sympify(formula_str) 
    integral = sp.integrate(f, x) 
    return integral + c1
```

& M_1(2) - M_2(0) = 0 \hspace{10pt}\Rightarrow\hspace{10pt} -\frac{2^4}{2} + A\cdot2 - 4\frac{0^3}{3} + 12\cdot0^2 - D_1\cdot0 - D_2 = 0 \hspace{5pt}\Rightarrow\hspace{5pt} D_2 = 2 \cdot A - 8\\\\
& Q_2(3) + B = 0 \hspace{10pt}\Rightarrow\hspace{10pt} 4 \cdot 3^2 - 24 \cdot 3 + D_1 + B = 0 \hspace{5pt}\Rightarrow\hspace{5pt} B = 36 - D_1\\\\
& M_2(3) = 0 \hspace{10pt}\Rightarrow\hspace{10pt} 4\cdot \frac{3^3}{3} - 12 \cdot 3^2 + 3\cdot (A - 31) + 2 \cdot A - 8 = 0\hspace{5pt}\Rightarrow\hspace{5pt} A = 34.6\\\\
& A = C_1 = 34.6\ kN\\\\
& C_2 = 0\\\\
& D_1 = 3.6\ kN\\\\
& D_2 = 61.2\ kN\\\\
& B = 32.4\ kN
\end{align}$$

From this it's clear that we don't need to calculate the spring reactions in advance, we can gain them from the above 6 equations. 

```python
""" 
Defining Spring reactions and Integral constants as symbols
"""

x, C_1, C_2, D_1, D_2 = sp.symbols("x C_1 C_2 D_1 D2")  # Integral constants
A, B = sp.symbols("A B")                                # Spring reactions

"""
Calculate quadratic equation coefficients and quadratic load equation for the first section
"""
a_1, b_1, c_1 = calculate_quadratic_eq_coefficients(0, q, 0, l_1)
quadratic_equation = a_1 * x**2 + b_1 * x + c_1

"""
Calculate linear equation cofficients and linear load equation for the second section:
"""
a_2, b_2 = calculate_linear_eq_coefficients(0, q, l_2, 0)
linear_equation = a_2 * x + b_2

"""
We consider the first section very close to the beginning. The distributed load is a parabolic function.
As x ~ 0 we substitude it 0.
We don't have any axial forces, so we only have 2 equations for equilibrium:
Sum of all the vertically acting forces and sum of all the moments acting on that section
"""

"""
Sum vertical forces = 0
"""
v_1 = - integrate_function(str(quadratic_equation)) + C_1
v_1_at_0 = v_1.subs(x, 0)

"""
First equation:
"""
v_1_equation = sp.Eq(v_1_at_0 - A, 0)

"""
Sum moments = 0:
"""
m_1 = integrate_function(str(v_1)) + C_2
m_1_at_0 = m_1.subs(x, 0)

"""
Second equation:
"""
m_1_equation = sp.Eq(m_1_at_0, 0)

"""
We consider the end of the first section (x = l_1) and start of the second section (x = 0).
Here ends the parabolic distributed load, starts the linear distributed load, and also there is a point load with
intensity = f
"""

"""
Sum vertical forces = 0
"""
v_1_boundry_at_l_1 = v_1.subs(x, l_1)
v_2 = - integrate_function(str(linear_equation)) + D_1
v_2_boundry_at_0 = v_2.subs(x, 0)

"""
Third equation:
"""
v_boundry_equation = sp.Eq(v_1_boundry_at_l_1 - f - v_2_boundry_at_0, 0)

"""
Sum moments = 0:
"""
m_1_boundry = m_1.subs(x, l_1)
m_2 = integrate_function(str(v_2)) + D_2
m_2_boundry_at_0 = m_2.subs(x, 0)

"""
Fourth equation:
"""
m_boundry_equation = sp.Eq(m_1_boundry - m_2_boundry_at_0, 0)

"""
Here we consider the end of the second section - x = l_2 and the distributed load is a linear function.
"""

"""
Sum vertical forces = 0
"""
v_2_at_l_2 = v_2.subs(x, l_2)
"""
Fifth equation:
"""
v_2_equation = sp.Eq(v_2_at_l_2 + B, 0)

"""
Sum moments = 0:
"""
m_2_at_l_2 = m_2.subs(x, l_2)

"""
Sixth equation:
"""
m_2_equation = sp.Eq(m_2_at_l_2, 0)

"""
Solving the six equations
"""
eqs = [
    v_1_equation,m_1_equation,
      v_boundry_equation, m_boundry_equation, 
       v_2_equation, m_2_equation
]

result = solve(eqs,(A, B, C_1, C_2, D_1, D_2))
```

 - Draw the diagrams of Internal forces:

```python
def draw_integral_internal_force_diagram(x_1_array, y_1_array, x_2_array, y_2_array, force_label, diagram_title, 
                                         units, invert_y = False):
    
    max_internal_force_1 = np.max(y_1_array)
    max_internal_force_1_location = x_1_array[np.argmax(y_1_array)]
    min_internal_force_1 = np.min(y_1_array)
    min_internal_force_1_location = x_1_array[np.argmin(y_1_array)]

    max_internal_force_2 = np.max(y_2_array)
    max_internal_force_2_location = x_2_array[np.argmax(y_2_array)]
    min_internal_force_2 = np.min(y_2_array)
    min_internal_force_2_location = x_2_array[np.argmin(y_2_array)]
    
    plt.plot(x_1_array, y_1_array, color = "blue")
    plt.plot(x_2_array, y_2_array, color="blue")
    plt.axhline(0, color='black', lw=1)
    plt.fill_between(x_1_array, y_1_array, alpha=0.3, color = "blue")
    plt.fill_between(x_2_array, y_2_array, alpha=0.3, color= "blue")
    
    plt.scatter(max_internal_force_1_location, max_internal_force_1, color='red', 
                label=f'{force_label}max={max_internal_force_1:.2f}')
    plt.annotate(f"{force_label}max = {max_internal_force_1:.2f} {units}", 
             xy=(max_internal_force_1_location, max_internal_force_1), 
             xytext=(max_internal_force_1_location - 2, max_internal_force_1 + 1))

    plt.scatter(max_internal_force_2_location, max_internal_force_2, color='red', 
                label=f'{force_label}max={max_internal_force_2:.2f}')
    plt.annotate(f"{force_label}max = {max_internal_force_2:.2f} {units}", 
             xy=(max_internal_force_2_location, max_internal_force_2), 
             xytext=(max_internal_force_2_location + 0.5, max_internal_force_2 + 1))

    plt.scatter(min_internal_force_1_location, min_internal_force_1, color='red', 
                label=f'{force_label}min={min_internal_force_1:.2f}')
    plt.annotate(f"{force_label}min = {min_internal_force_1:.2f} {units}", 
             xy=(min_internal_force_1_location, min_internal_force_1), 
             xytext=(min_internal_force_1_location +0.5, min_internal_force_1 -0.5))

    plt.scatter(min_internal_force_2_location, min_internal_force_2, color='red', 
                label=f'{force_label}min={min_internal_force_2:.2f}')
    plt.annotate(f"{force_label}min = {min_internal_force_2:.2f} {units}", 
             xy=(min_internal_force_2_location, min_internal_force_2), 
             xytext=(min_internal_force_2_location + 0.5, min_internal_force_2 -0.5))

    if invert_y:
        plt.gca().invert_yaxis()
    
    plt.title(diagram_title)
    plt.xlabel("L, m")
    plt.ylabel(f"{force_label}, {units})
    plt.grid(True)
    plt.show()
```

 - Draw Shear force diagram withe extremum values:

```python
draw_integral_internal_force_diagram(x_1, shear_first, (x_2 + l_1), shear_second, "Q", "SHEAR FORCES", "kN", False)
```
